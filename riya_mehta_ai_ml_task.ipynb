```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import whisper
import numpy as np
import re
file_path = input("enter path")

y, sr = librosa.load(file_path, sr=16000)
duration = librosa.get_duration(y=y, sr=sr)

print(f"Loaded {file_path} | Duration: {round(duration, 2)} sec")
```

```python
plt.figure(figsize=(12, 4))
librosa.display.waveshow(y, sr=sr)
plt.title("Waveform")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Detect non-silent segments
intervals = librosa.effects.split(y, top_db=20)
speech_time = sum((end - start) for start, end in intervals) / sr
pause_ratio = (duration - speech_time) / duration

print(f"Pause Ratio: {round(pause_ratio, 2)}")
```

```python
model = whisper.load_model("base")
result = model.transcribe(file_path)
transcript = result["text"].lower()

print("\nTranscript:")
print(transcript)
```

```python
# Common filler and confusion phrases
filler_phrases = [
    " uh ", " um ", " er ", " hmm ",
    "word for", "what do you call", "the thing that"
]

# Count filler phrases
filler_count = sum(transcript.count(p) for p in filler_phrases)

# Count repeated words (e.g., "I, I went...")
words = re.findall(r'\b\w+\b', transcript)
repeated_words = sum(1 for i in range(1, len(words)) if words[i] == words[i-1])

# Count pause markers (e.g., "...", "--")
pause_marks = ["...", "--"]
pause_count = sum(transcript.count(p) for p in pause_marks)

# Speech rate (words per second)
word_count = len(words)
speech_rate = word_count / duration
```

```python
print("\n--- Feature Summary ---")
print(f"Duration: {round(duration, 2)} s")
print(f"Speech Rate: {round(speech_rate, 2)} words/sec")
print(f"Filler Word Count: {filler_count}")
print(f"Repeated Words: {repeated_words}")
print(f"Pause Markers: {pause_count}")
print(f"Pause Ratio: {round(pause_ratio, 2)}")
```
